# Student Management System :

```python
students = []

for i in range(3):
    student = input("Enter the Student name:").strip().title()
    subject = input("Enter the subject:").strip().title()
    marks = int(input("Enter the marks:"))
    students.append(student)
    students.append(subject)
    students.append(marks)
    
    ## For updating the names:
for i in range(3):
    update_name = input("Do you want to update the student name:(YES/NO)")
    if update_name == "YES":
        old_name = input("Enter the student name you want to update:").strip().title()
        if old_name in students:
            index = students.index(old_name)
            new_name = input("Enter the new student name:").strip().title()
            students[index] = new_name
            print(students)
            print("------student name updated successfully------")
        else:
            print("student not found!")  
            
    elif old_name ==" NO":
        break

 ## for updating the subjects:
for i in range(3):
     update_subject = input("DO you want to update the any student subject:(YES/NO)")
     if update_subject == "YES":
          old_subject = input("Enter the suject name you want to update:").strip().title()
          if old_subject in students:
              index = students.index(old_subject)
              new_subject = input("Enter the new subject name:").strip().title()
              students[index] = new_subject
              print(students)
              print("-------Subject updated successfully-------") 
          else:
               print("Subject not found!")   
        
              
     elif old_subject == "NO":
         break
          
        
    ### For updating marks:     
         
     

for i in range(3):
     update_marks = (input("DO you want to update the any student marks:(YES/NO)"))
     if update_marks == "YES":
          old_marks = int(input("Enter marks you want to update:")).strip().title()
          if old_marks in students:
              index = students.index(old_marks)
              new_marks = int(input("Enter the new marks:"))
              students[index] = new_marks
              print(students)
              print("-------marks updated successfully-------")   
          
          else:
               print("marks not found!")
          
     elif update_marks == "NO":
         break    
          
    
        
    ### For deleting any student from students:
    
for i in range(3):
    delete = input("Do you want to delete any student from list:(YES/NO)")
    if delete == "YES":
        delete_student = input("Enter the student name:").strip().title()
        if delete_student in students:
            index = students.index(delete_student)
            del students[index:index + 3]
            print(students)
            print("-----Student record deleted successfully-----")
        
        else:
            print("Student not found!")
       
        
    
    elif delete == "NO":
        break
        
   ## for searching student exist or not in students:
   
    
for i in range(3):
    search = input("Do you want to search any student record:(YES/NO)")
    if search == "YES":
        search_name = input("Enter the stundent name:").strip().title()
        if search_name in students:
            index = students.index(search_name)
             
            print("-----you search this student-----")
            print(students[index:index + 3])
        else:
            print("student not found!")                 
    
   
    elif search == "NO":
        break
         
    ### for seeing all students:     
                    
for student in students:
    print(student)    
           
            
marks_list = students[2::3]

avg_marks = sum(marks_list) / len(marks_list)

print("Average Marks:", avg_marks)
    
```
