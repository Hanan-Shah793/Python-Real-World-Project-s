# Student Management System :

```python
print("-----Student Management System-----")
students = []

def add_student():
    name = input("Enter the student name:").strip().title()
    subject = input("Enter the subject name:").strip().title()
    marks = int(input("Enter the marks:"))
    
    student = {
        "name" : name,
        "subject" : subject,
        "marks" : marks
    }
    
    students.append(student)
    print("-----\n Student added successfully-----")
    
def view_student():
    if students:
        for student in students:
            print("------------------------")
            print("Student Name:", student["name"])
            print("Subject:", student["subject"])
            print("Marks:", student["marks"])
            print("------------------------")
    else:
        print("NO Student Record Found!!")
    
        
   
def update_student():
    name = input("Enter the student name for updation:").strip().title()
    for student in students:
        if student["name"] == name :
            student["name"] = input("Enter new name").strip().title()
            student["subject"] = input("Enter new subject").strip().title()
            student["marks"] = int(input("Enter new marks"))
            print("----Student Updated!!----")
            return
        
    print("Student Not Found!!")
    

def delete_student():
    name = input("Enter the student name you want too delete them:").strip().title()
    for student in students:
        if student["name"] == name:
            students.remove(student)
            print("-----Student Deleted-----")
            return
    print("Student Not Found!!!")
    
def search_student():
    name = input("Enter the student name for checking:").strip().title()
    for student in students:
        if student["name"] == name:
            print(student)
            return
    print("Student Not Found!!!")
    

def average_marks():
    if students:
        total = 0
        for student in students:
            total += student["marks"]
        print("average of marks:" , total / len(students)) 
    else:
        print("Student Not Found!!!")
            
            
def class_topper():
    if students:
        topper = students[0]
        for student in students:
            if student['marks'] > topper["marks"]: 
                topper = student       
                
        print("Class Topper:" , topper["name"])
        print("topper Marks:" , topper['marks'])
    else:
        print("No Record Found!!!")
            

def grade_system():
    if not students:
        print("No Record Found!!")
        return
    for student in students:
        mark = student['marks']
        
        if mark >=90:
            grade = "A+"
        elif mark >= 80:
            grade = "A"
        elif mark >= 70:
            grade = "B"
        elif mark >= 60:
            grade = "C" 
        else:
            grade = "Need Improvement!!!"
            
        if mark >= 50:
            status = "Pass"
        else:
            status = "fail"
            
        print("------------------------")
        print("Student Name:", student["name"])
        print("Subject:", student["subject"])
        print("Marks:", mark)
        print("Grade:", grade)
        print("Status:", status)
        print("------------------------")
        

def subject_wise_search():
    subject = input("Enter the subject name:").strip().title()
    found = False
    for student in students:
        if student["subject"] == subject:
            print(student)
            found = True
            
    if not found:
        print("Subject Not Found!!!") 
    
   
    
while True:
    print("""
        1 add_student
        2 view_student
        3 update_student
        4 delete_student
        5 search_student
        6 average_marks
        7 class_topper
        8 grade_system
        9 subject_wise_search
        0 Exit
    """)
    choice = input("Enter you choice:").strip()
    if choice == "1":
        add_student()
    elif choice == "2":
        view_student()
    elif choice == "3":
        update_student()
    elif choice == "4":
        delete_student()
    elif choice == "5":
        search_student()
    elif choice == "6":
        average_marks()
    elif choice == "7":
        class_topper()
    elif choice == "8":
        grade_system()
    elif choice == "9":
        subject_wise_search()
    elif choice == "0":
        print("------Program_Closed!!------")
        break
    else:
        print("-----Invalid Choice!!------")
```
